# Softmax Regression (Multinomial Logistic Regression)

Softmax Regression is a generalization of logistic regression that we can use for multi-class classification (under the assumption that the classes are mutually exclusive). Instead of returning a single probability for a binary outcome, the softmax function outputs a probability distribution across $K$ possible classes.
## 1. The Softmax Function from Scratch

The softmax function takes a vector $z$ of $K$ real numbers and normalizes it into a probability distribution consisting of $K$ probabilities proportional to the exponentials of the input numbers.

Formula:
$$ \sigma(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}} $$

Let's implement this mathematical function using NumPy.

```python
import numpy as np

def softmax(z):
    """
    Computes the softmax probability distribution for a given array of scores.
    """
    probability = np.exp(z) / np.sum(np.exp(z))
    return probability

# Test the function with a sample array
arr = [2, 5, -1]
sm = softmax(arr)

print("Raw Scores (Logits):", arr)
print("Softmax Probabilities:", sm)
```

## 2. Softmax Regression using Scikit-Learn

In `scikit-learn`, Softmax Regression is implemented via the `LogisticRegression` class by setting the `multi_class` parameter to `'multinomial'`.

We will train this model on the **Iris dataset**, a classic multi-class classification dataset with 3 different classes of iris flowers.

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

# 1. Load the Iris dataset
X, y = load_iris(return_X_y=True)

# 2. Initialize the model
# multi_class='multinomial' triggers Softmax Regression
# max_iter is increased to 200 to ensure the solver converges
model = LogisticRegression(multi_class='multinomial', max_iter=200)

# 3. Fit the model to the data
model.fit(X, y)

# 4. Predict probabilities for the first 5 samples
y_pred_proba = model.predict_proba(X[:5])
print("Predicted Probabilities for first 5 samples:\n", y_pred_proba)

# 5. Extract the final class predictions using np.argmax (highest probability)
y_pred_classes = np.argmax(y_pred_proba, axis=1)

print("\nPredicted Classes:", y_pred_classes)
print("Actual Classes:   ", y[:5])
```

### Summary of Results:
* The `predict_proba` method outputs an array where each row sums to 1 (just like our custom softmax function).
* `np.argmax(y_pred_proba, axis=1)` selects the index (class) with the highest probability for each row.
* By comparing the predicted classes to the actual classes (`y[:5]`), we can see that our model correctly predicted the first 5 samples.
